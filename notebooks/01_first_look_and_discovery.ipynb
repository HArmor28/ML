## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

```python
# Your discovery here

# Considering pages with at least 100 impressions, the results find the content_type with the worst CTR within each position tier. This analysis shows that compairson article is generally the poorest performer across several position tiers, while feedly article has the lowest CTR overall, in the deep tier. These findings are visualized by the bar chart below.
# Filter for visible pages (impressions >= 100) as in Discovery B
visible = df[df["impressions_90d"] >= 100]

# Group by position_tier and content_type and calculate the mean CTR
ctr_by_pos_content = visible.groupby(["position_tier", "content_type"])[["ctr"]].mean()

# Find the content_type with the worst (lowest) CTR for each position_tier
worst_ctr_content_type = ctr_by_pos_content.groupby(level=0).apply(lambda x: x.idxmin())

print("Content type with the worst CTR for each position tier (impressions >= 100):")
# Extract just the content_type for clearer display
# The result is a Series of tuples, so we need to unpack the content_type from the second element of the tuple
display(worst_ctr_content_type.apply(lambda x: x[1]))

print("\nFull breakdown of mean CTR by position tier and content type:")
display(ctr_by_pos_content.unstack().round(4))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a figure and axes for the plot
fig, ax = plt.subplots(figsize=(12, 7))

# Unstack the ctr_by_pos_content DataFrame to easily plot content_types by position_tier
ctr_by_pos_content.unstack().plot(kind='bar', ax=ax)

# Add title and labels for clarity
ax.set_title('Mean CTR by Position Tier and Content Type (Impressions >= 100)')
ax.set_xlabel('Position Tier')
ax.set_ylabel('Mean CTR')
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
ax.legend(title='Content Type')
plt.tight_layout()
plt.show()
```
